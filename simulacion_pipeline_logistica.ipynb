```python
#Automatización y Resumen Financiero por Zonas, en una Empresa de Logística

#Configuración y Conexión (Las Raíces)
import sqlalchemy
import pandas as pd
import numpy as np

def conectar_base_datos():
  db_url="postgresql://usuario:contraseña@localhost:5432/bd_logistica_bogota"
  engine=sqlalchemy.create_engine(db_url)
  return engine

#Módulo de Extracción (Las Ramas)

def extraer_datos_sql(engine):
  query="""
  SELECT
    id_transaccion,
            tienda,
            monto,
            fecha_transaccion,
            estado
    FROM
      transacciones_bogota
    WHERE
      estado ='Completado' AND tienda ILIKE '%Norte%'
    """
    df=pd.read_sql(query, engine)
    return df

    #Ramas de Limpieza y Transformación (Pandas)
    def limpiar_y_procesar_datos(df):
      #Limpieza de Datos Duplicados y Nulos
      df=df.drop_duplicates(subset=['id_transaccion'])
      df['monto']=df.groupby('tienda')['monto'].transform(lambda x: x.fillna(x.mean))
      #Tranformación: Detectar anomalias
      umbral_riesgo = 5000000.0
      df['es_anomalia'] = df['monto'] > umbral_riesgo

    return df

    #Los Frutos (resumen final)

    def generar_reporte_ejecutivo(df):
      resumen=df.groupby('tienda').agg(total_transacciones=('id_transaccion', 'count'),
                                       monto_total_recaudado=('monto', 'sum'),
                                       total_anomalias_detectadas=('es_anomalia', 'sum')).
                                       reset_index()
        return resumen

      #Tronco Principal (Orquestación)
      if__name__=="__main__":
        print("Iniciando pipeline de procesamiento para el cliente...")
        # 1. Conectar
    engine_conexion = conectar_base_datos()
    # 2. Extraer
    df_crudo = extraer_datos_sql(engine_conexion)
    print(f"Datos extraídos con éxito: {len(df_crudo)} registros encontrados.")
    # 3. Procesar y Limpiar
    df_procesado = limpiar_y_procesar_datos(df_crudo)
    # 4. Generar Reporte
    reporte_final = generar_reporte_ejecutivo(df_procesado)

    # 5. Exportar entregable (Lo que el cliente puede pagar y descargar)
    ruta_salida = "reporte_auditoria_norte.csv"
    reporte_final.to_csv(ruta_salida, index=False)

    print(f"Proceso finalizado. Entregable guardado exitosamente en: {ruta_salida}")
```
