```r
# Project 3: Time-Series Modeling & Reproducible Workflows
# Simulating historical landings time-series data for environmental monitoring

set.seed(55)
years <- 2010:2025
landings <- 5000 + (years - 2010) * 120 + rnorm(length(years), mean = 0, sd = 450)
df_ts <- data.frame(Year = years, Total_Landings_Ton = landings)

print("Historical Landings Summary:")
summary(df_ts$Total_Landings_Ton)
```

```r
# Cell 2: Annual growth and trend metrics
mean_landings <- mean(df_ts$Total_Landings_Ton)
max_landings <- max(df_ts$Total_Landings_Ton)
min_landings <- min(df_ts$Total_Landings_Ton)

print(paste("Mean Annual Landings:", round(mean_landings, 2), "Tons"))
print(paste("Peak Landings Recorded:", round(max_landings, 2), "Tons"))
print(paste("Minimum Landings Recorded:", round(min_landings, 2), "Tons"))
```

```r
# Cell 3: Time-series plot with smoothed moving average trend
plot(df_ts$Year, df_ts$Total_Landings_Ton, type = "b", pch = 18, col = "darkgreen", lwd = 2,
     main = "Historical Landings Time-Series & Trend Analysis (2010-2025)",
     xlab = "Year", ylab = "Total Landings (Tons)")

# Adding smoothed trend line
lines(df_ts$Year, smooth(df_ts$Total_Landings_Ton), col = "orange", lwd = 2, lty = 2)
legend("bottomright", legend = c("Observed Landings", "Smoothed Trend"), col = c("darkgreen", "orange"), lty = c(1, 2), lwd = 2)
grid()
```
