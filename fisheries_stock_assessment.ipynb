```r
# Create a simulated dataframe of fishery resources
fishery_data <- data.frame(
  Species = c("Hake", "Tuna", "Sardine", "Cod"),
  Estimated_Biomass = c(1250, 840, 3100, 1600)
)

# Display a basic statistical summary
print("Biomass Summary by Species:")
print(fishery_data)

# Generate a basic integrated bar chart for your portfolio
barplot(fishery_data$Estimated_Biomass,
        names.arg = fishery_data$Species,
        col = "steelblue",
        main = "Preliminary Biomass Assessment",
        ylab = "Biomass (tons)",
        xlab = "Species")
```

```r
# Biomass Assessment & Fishery Analysis
set.seed(42)
years <- 2015:2025
biomass <- 12000 + cumsum(rnorm(length(years), mean = 150, sd = 800))
df_biomass <- data.frame(Year = years, Biomass_Ton = biomass)

print("Biomass Statistical Summary:")
summary(df_biomass$Biomass_Ton)

plot(df_biomass$Year, df_biomass$Biomass_Ton, type = "o", col = "blue", pch = 16, lwd = 2,
     main = "Exploited Biomass Assessment (2015-2025)",
     xlab = "Year", ylab = "Estimated Biomass (Tons)")
grid()
```

```r
# Cell 3: Advanced Fishery Indicator - Exploitation Rate Simulation
# Calculating relative exploitation metrics for scientific advisory reporting

set.seed(42)
years <- 2015:2025
landings <- 2500 + rnorm(length(years), mean = 100, sd = 300)
estimated_biomass <- 12000 + cumsum(rnorm(length(years), mean = 150, sd = 800))

# Compute Exploitation Rate (Catch / Biomass)
exploitation_rate <- landings / estimated_biomass
df_indicators <- data.frame(Year = years, Landings = landings, Exploitation_Rate = exploitation_rate)

print("Summary of Exploitation Rate Indicators:")
summary(df_indicators$Exploitation_Rate)

# Plotting the exploitation rate trend
plot(df_indicators$Year, df_indicators$Exploitation_Rate, type = "b", col = "darkred", pch = 17, lwd = 2,
     main = "Simulated Stock Exploitation Rate Trend (2015-2025)",
     xlab = "Year", ylab = "Exploitation Rate (Catch/Biomass)")
abline(h = mean(df_indicators$Exploitation_Rate), col = "gray", lty = 2, lwd = 2)
legend("topright", legend=c("Exploitation Rate", "Mean Level"), col=c("darkred", "gray"), lty=c(1,2), lwd=2)
grid()
```
