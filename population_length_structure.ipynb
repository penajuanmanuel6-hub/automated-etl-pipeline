```r
# Project 2: Population Structure & Length-Frequency Analysis
# Simulating length-frequency distribution for population dynamics assessment

set.seed(101)
lengths <- rnorm(n = 1200, mean = 48, sd = 7)
df_lengths <- data.frame(Total_Length_cm = lengths)

print("Summary of Population Lengths:")
summary(df_lengths$Total_Length_cm)
```

```r
# Cell 2: Descriptive metrics for population modelling
mean_length <- mean(df_lengths$Total_Length_cm)
median_length <- median(df_lengths$Total_Length_cm)
sd_length <- sd(df_lengths$Total_Length_cm)

print(paste("Mean Length:", round(mean_length, 2), "cm"))
print(paste("Median Length:", round(median_length, 2), "cm"))
print(paste("Standard Deviation:", round(sd_length, 2), "cm"))
```

```r
# Cell 3: Length-frequency histogram with reference limits
hist(df_lengths$Total_Length_cm, breaks = 25, col = "steelblue", border = "white",
     main = "Population Length-Frequency Distribution in Catch",
     xlab = "Total Length (cm)", ylab = "Absolute Frequency")

# Adding vertical line for mean reference size
abline(v = mean_length, col = "red", lwd = 2, lty = 2)
legend("topright", legend = c("Mean Length"), col = c("red"), lty = 2, lwd = 2)
grid()
```
