```python
import numpy as np
import pandas as pd

# 1. Simulación de datos de partidas multijugador
np.random.seed(101)
partidas_df = pd.DataFrame({
    'servidor_region': np.repeat(['Bogota_East', 'Medellin_Core', 'Lima_Hub'], 40),
    'tiempo_cola_seg': np.random.exponential(scale=15, size=120),
    'k_d_ratio': np.random.normal(loc=1.1, scale=0.4, size=120).clip(0.1, 4.0)
})

# 2. Función modular para calcular indicadores estadísticos limpios por región
def calcular_metricas_servidor(df, grupo_col, metrica_cola, metrica_kd):
    resumen = df.groupby(grupo_col).agg(
        total_partidas=('tiempo_cola_seg', 'count'),
        cola_promedio_seg=(metrica_cola, lambda x: round(x.mean(), 1)),
        cola_mediana_seg=(metrica_cola, lambda x: round(x.median(), 1)),
        kd_medio=(metrica_kd, lambda x: round(x.mean(), 2)),
        kd_maximo=(metrica_kd, lambda x: round(x.max(), 2))
    ).reset_index()
    return resumen

# 3. Ejecución del procesamiento analítico entregable
df_indicadores = calcular_metricas_servidor(
    partidas_df, 'servidor_region', 'tiempo_cola_seg', 'k_d_ratio'
)

print("--- RESUMEN TÉCNICO DE RENDIMIENTO DE SERVIDORES ---")
print(df_indicadores)

# El entregable se exporta limpiamente para el equipo de desarrollo
# df_indicadores.to_csv("reporte_rendimiento_servidores.csv", index=False)
```
